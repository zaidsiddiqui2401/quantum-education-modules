# GHZ State

The Greenberger-Horne-Zeilinger (GHZ) state is a maximally entangled state, just like the Bell states. It represents one of the simplest examples of entanglement with more than $2$ qubits in quantum mechanics.

## GHZ State 

### Question 1  
What is the final state of the 3-qubit GHZ circuit?

A) \(|000\rangle\)  
B) \(|111\rangle\)  
C) \(\frac{1}{\sqrt{2}} (|000\rangle + |111\rangle)\)  
D) \(\frac{1}{\sqrt{2}} (|001\rangle + |110\rangle)\)  

**Answer:** C

---

### Question 2  
Which sequence of gates creates a GHZ state?

A) X on all qubits  
B) H on q[0], then CX(q[0]→q[1]), CX(q[0]→q[2])  
C) CX(q[1]→q[0]), then H on q[2]  
D) Z on all qubits  

**Answer:** B

---

### Question 3  
What measurement outcomes are expected?

A) All 8 states equally likely  
B) Only 000  
C) Only 111  
D) 000 and 111 only  

**Answer:** D

---

### Question 4  
What is the probability of measuring each outcome (000 or 111)?

A) 0%  
B) 25%  
C) 50%  
D) 100%  

**Answer:** C

---

### Question 5  
What do the measurement results demonstrate?

A) Independent qubits  
B) Measurement error  
C) Strong entanglement (all qubits correlated)  
D) Classical randomness  

**Answer:** C

```python
import ipywidgets as widgets
from IPython.display import display, HTML

def make_question(q_num, question, options, correct):
    
    question_html = widgets.HTML(
        value=f"""
        <div style="color: black; font-size: 16px; line-height: 1.6;">
            <h4 style="margin-bottom: 10px;">Q{q_num}: {question}</h4>
            {''.join([f"<div>{opt}</div>" for opt in options])}
        </div>
        """
    )

    radio = widgets.RadioButtons(
        options=["A", "B", "C", "D"],
        value=None,
        description="Select:",
        style={"description_width": "initial"},
        layout=widgets.Layout(width="120px")
    )

    button = widgets.Button(
        description="Check Answer",
        button_style="info"
    )

    output = widgets.Output()

    def check_answer(b):
        output.clear_output(wait=True)
        with output:
            if radio.value is None:
                print("Select an answer first.")
            elif radio.value == correct:
                print("Correct!")
            else:
                print("Incorrect. Try again.")

    button.on_click(check_answer)

    display(question_html, radio, button, output)


make_question(
    1,
    "What is the final state of the 3-qubit GHZ circuit?",
    [
        "A. |000⟩",
        "B. |111⟩",
        "C. (|000⟩ + |111⟩)/√2",
        "D. (|001⟩ + |110⟩)/√2"
    ],
    "C"
)

make_question(
    2,
    "Which sequence of gates creates a GHZ state?",
    [
        "A. X on all qubits",
        "B. H on q[0], then CX(q[0]→q[1]), CX(q[0]→q[2])",
        "C. CX(q[1]→q[0]), then H on q[2]",
        "D. Z on all qubits"
    ],
    "B"
)

make_question(
    3,
    "What measurement outcomes are expected?",
    [
        "A. All 8 states equally likely",
        "B. Only 000",
        "C. Only 111",
        "D. 000 and 111 only"
    ],
    "D"
)

make_question(
    4,
    "What is the probability of measuring each outcome (000 or 111)?",
    [
        "A. 0%",
        "B. 25%",
        "C. 50%",
        "D. 100%"
    ],
    "C"
)

make_question(
    5,
    "What do the measurement results demonstrate?",
    [
        "A. Independent qubits",
        "B. Measurement error",
        "C. Strong entanglement (all qubits correlated)",
        "D. Classical randomness"
    ],
    "C"
)
```
